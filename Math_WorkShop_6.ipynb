```python
def matrix_matrix_product(matrix1, matrix2):
    matrix1 = np.asarray(matrix1)
    matrix2 = np.asarray(matrix2)

    rows1, columns1 = matrix1.shape
    rows2, columns2 = matrix2.shape

    if columns1 != rows2:
        raise ValueError("Matrices are not compatible for multiplication.")

    result = np.zeros((rows1, columns2), dtype=float)

    for i in range(rows1):
        for j in range(columns2):
            result[i, j] = dot_product(
                matrix1[i, :],
                matrix2[:, j]
            )

    return result


A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

B = np.array([
    [7, 8],
    [9, 10],
    [11, 12]
])

manual_result = matrix_matrix_product(A, B)
numpy_result = A @ B

print("Matrix A =")
print(A)

print("Matrix B =")
print(B)

print("Manual Result =")
print(manual_result)

print("NumPy Result =")
print(numpy_result)
```

```python
def is_symmetric(matrix):
    matrix = np.asarray(matrix)

    # A symmetric matrix must be square
    if matrix.ndim != 2 or matrix.shape[0] != matrix.shape[1]:
        return False

    return np.allclose(matrix, matrix.T)


matrix1 = np.array([
    [1, 2, 3],
    [2, 4, 5],
    [3, 5, 6]
])

matrix2 = np.array([
    [1, 2],
    [3, 4]
])

print("Matrix 1 is symmetric:", is_symmetric(matrix1))
print("Matrix 2 is symmetric:", is_symmetric(matrix2))
```

Task 2: Bike details in line plot

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]

enfield = [50, 40, 70, 80, 20]
honda = [80, 20, 20, 50, 60]
yamaha = [70, 20, 60, 40, 60]
ktm = [40, 20, 30, 70, 60]

plt.plot(days, enfield, marker='o', label='Enfield')
plt.plot(days, honda, marker='o', label='Honda')
plt.plot(days, yamaha, marker='o', label='Yamaha')
plt.plot(days, ktm, marker='o', label='KTM')

plt.title("Bike Details in Plot")
plt.xlabel("Days")
plt.ylabel("Distance Covered in KMs")

plt.legend()
plt.grid()
plt.show()
```

Exercise 3: Plot 20 values from the given functions
Function 1: f(x) = 5x³ + 2x − 1

```python
import matplotlib.pyplot as plt

x_values = []
y_values = []

for x in range(1, 21):
    y = 5 * x**3 + 2 * x - 1
    x_values.append(x)
    y_values.append(y)

plt.plot(x_values, y_values, marker='o')

plt.title("Graph of f(x) = 5x³ + 2x - 1")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.grid()
plt.show()
```

Function 2: f(x) = -2x³ + x² + 100

```python
import matplotlib.pyplot as plt

x_values = []
y_values = []

for x in range(1, 21):
    y = -2 * x**3 + x**2 + 100
    x_values.append(x)
    y_values.append(y)

plt.plot(x_values, y_values, marker='o')

plt.title("Graph of f(x) = -2x³ + x² + 100")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.grid()
plt.show()
```

Function 3: f(x) = 2πx + 20

```python
import matplotlib.pyplot as plt
import math

x_values = []
y_values = []

for x in range(1, 21):
    y = 2 * math.pi * x + 20
    x_values.append(x)
    y_values.append(y)

plt.plot(x_values, y_values, marker='o')

plt.title("Graph of f(x) = 2πx + 20")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.grid()
plt.show()
```

Exercise 4: Plot all graphs from Question 3 using subplot

```python
import matplotlib.pyplot as plt
import math

x_values = list(range(1, 21))

y1 = []
y2 = []
y3 = []

for x in x_values:
    y1.append(5 * x**3 + 2 * x - 1)
    y2.append(-2 * x**3 + x**2 + 100)
    y3.append(2 * math.pi * x + 20)

plt.subplot(2, 2, 1)
plt.plot(x_values, y1, marker='o')
plt.title("f(x) = 5x³ + 2x - 1")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid()

plt.subplot(2, 2, 2)
plt.plot(x_values, y2, marker='o')
plt.title("f(x) = -2x³ + x² + 100")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid()

plt.subplot(2, 2, 3)
plt.plot(x_values, y3, marker='o')
plt.title("f(x) = 2πx + 20")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid()

plt.tight_layout()
plt.show()
```

Exercise 5: Plot all graphs from Task 2 using subplot

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]

enfield = [50, 40, 70, 80, 20]
honda = [80, 20, 20, 50, 60]
yamaha = [70, 20, 60, 40, 60]
ktm = [40, 20, 30, 70, 60]

plt.subplot(2, 2, 1)
plt.plot(days, enfield, marker='o')
plt.title("Enfield")
plt.xlabel("Days")
plt.ylabel("Distance in KMs")
plt.grid()

plt.subplot(2, 2, 2)
plt.plot(days, honda, marker='o')
plt.title("Honda")
plt.xlabel("Days")
plt.ylabel("Distance in KMs")
plt.grid()

plt.subplot(2, 2, 3)
plt.plot(days, yamaha, marker='o')
plt.title("Yamaha")
plt.xlabel("Days")
plt.ylabel("Distance in KMs")
plt.grid()

plt.subplot(2, 2, 4)
plt.plot(days, ktm, marker='o')
plt.title("KTM")
plt.xlabel("Days")
plt.ylabel("Distance in KMs")
plt.grid()

plt.tight_layout()
plt.show()
```

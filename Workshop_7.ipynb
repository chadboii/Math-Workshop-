Part 1: Create and Understand Linear Systems:
Example 1: We will start with a simple problem. Consider the following linear systems:
a. 𝑥 + 3𝑦 = 5
𝑥 − 2𝑦 =− 5
𝑥 + 𝑦 = 3
b. 2𝑥 + 3𝑦 = 3
4𝑥 + 6𝑦 = 6
3𝑥 − 2𝑦 = 1
Solve the linear systems graphically.


```python
import numpy as np
import matplotlib.pyplot as plt
""" Part 1 - Problem 1: Solve linear systems graphically. """
# Generate 100 equidistant point between -5 to 5 for x
x = np.linspace(-5,5,100)
## Calculate y values for Equation 1 and 2
y1 = (5-x)/3
y2 = (5+x)/2
y3 = 3-x
## Generate a figure frame for plotting the lines
fig, ax = plt.subplots()
## Plot the two lines
ax.plot(x,y1, color='r')
ax.plot(x,y2, color='g')
ax.plot(x,y3, color='b')
## To decorate the plot, we insert equations.
ax.text(0.5,0.6, '$x+3y = 5$', color='r')
ax.text(-3,0.5,'$x-2y = -5$', color='g')
ax.text(1.2,2, '$x+y = 3$', color='b')
## Set the limits of the X & Y axes.
ax.set_xlim(-4,4)
ax.set_ylim(-2,6)
## Set property of X and Y axes, e.g., color and linewidth
ax.axvline(color='k',linewidth = 1)
ax.axhline(color='k',linewidth = 1)
## This option specifies the ticks based the list of numbers provided.
ax.set_xticks(list(range(-4,5)))
ax.set_aspect('equal')
ax.grid(True,ls=':')
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

Plot the graph for linear system b and write the point of intersection between three line

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 400)

# Rearrange each equation into y = ...
y1 = (3 - 2*x) / 3
y2 = (6 - 4*x) / 6
y3 = (3*x - 1) / 2

intersection_x = 9 / 13
intersection_y = 7 / 13

fig, ax = plt.subplots()

ax.plot(x, y1, label="2x + 3y = 3")
ax.plot(x, y2, linestyle="--", label="4x + 6y = 6")
ax.plot(x, y3, label="3x - 2y = 1")

ax.scatter(intersection_x, intersection_y)

ax.annotate(
    f"({intersection_x:.4f}, {intersection_y:.4f})",
    (intersection_x, intersection_y),
    xytext=(10, 10),
    textcoords="offset points"
)

ax.axhline(0, linewidth=1)
ax.axvline(0, linewidth=1)

ax.set_xlim(-4, 4)
ax.set_ylim(-4, 6)
ax.set_aspect("equal", adjustable="box")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_title("Linear System B")
ax.grid(True, linestyle=":")
ax.legend()

plt.show()
```

1. Create a program that reads in two 5x6 numpy.arrays, adds them together, and then stores
the result in a third 5x6 matrix. The program should then print out the first two matrices
and the result matrix.

```python
import numpy as np

def read_matrix(rows, columns, name):
    print(f"\nEnter {name} one row at a time.")
    print(f"Each row must contain {columns} numbers separated by spaces.")

    matrix = []

    for row_number in range(rows):
        while True:
            values = input(f"Row {row_number + 1}: ").split()

            if len(values) != columns:
                print(f"Enter exactly {columns} values.")
                continue

            try:
                row = [float(value) for value in values]
                matrix.append(row)
                break
            except ValueError:
                print("Invalid input. Enter numbers only.")

    return np.array(matrix)
```

2. Create a program that reads in three 2x3 numpy.arrays, subtracts the second from the
first, and adds the third to the result. The program should then print out all three matrices
and the result matrix.


```python
matrix_a = read_matrix(2, 3, "Matrix A")
matrix_b = read_matrix(2, 3, "Matrix B")
matrix_c = read_matrix(2, 3, "Matrix C")

result = matrix_a - matrix_b + matrix_c

print("\nMatrix A:")
print(matrix_a)

print("\nMatrix B:")
print(matrix_b)

print("\nMatrix C:")
print(matrix_c)

print("\nResult A - B + C:")
print(result)
```

3. Create a 4x4 matrix containing random numbers in the range of 1 to 20. Have the user
prompted for a number, check that the imputed value is a number and then check if the
number is in the matrix. If the number is in the matrix, return the number of times that
number is in the matrix.

```python
import numpy as np

rng = np.random.default_rng()

# 21 is excluded, so this generates numbers from 1 to 20.
matrix = rng.integers(1, 21, size=(4, 4))

print("Random 4x4 matrix:")
print(matrix)

while True:
    user_input = input("Enter an integer from 1 to 20: ")

    try:
        number = int(user_input)
        break
    except ValueError:
        print("Invalid input. Enter a whole number.")

count = np.count_nonzero(matrix == number)

if count > 0:
    print(f"{number} appears in the matrix {count} time(s).")
else:
    print(f"{number} is not present in the matrix.")
```

1. Create a matrix 𝑨 of order 5 × 5 with random entries. Calculate the rank of the matrix 𝑨
to check the number of linearly independent columns or rows and to check if the matrix is
invertible. What are the other ways to check if the matrix is invertible?


```python
import numpy as np

rng = np.random.default_rng(7)

A = rng.integers(1, 10, size=(5, 5))

rank_A = np.linalg.matrix_rank(A)
det_A = np.linalg.det(A)

print("Matrix A:")
print(A)

print("\nRank:", rank_A)
print("Determinant:", det_A)

if rank_A == A.shape[0]:
    print("Matrix A is invertible.")
    print("\nInverse:")
    print(np.linalg.inv(A))
else:
    print("Matrix A is not invertible.")
```

2. Generate a matrix 𝑨 of order 5 × 5 with random entries and generate a vector (column) 𝒙 of
length 5 with random entries. Calculate Matrix-vector product 𝒃 = 𝑨𝒙. Now, we have a linear
system, i.e. the matrix 𝑨 and a vector 𝒃. Solve this Linear System.


```python
import numpy as np

rng = np.random.default_rng(8)

A = rng.integers(1, 10, size=(5, 5))
original_x = rng.integers(1, 10, size=5)

b = A @ original_x

solved_x = np.linalg.solve(A, b)

print("Matrix A:")
print(A)

print("\nOriginal x:")
print(original_x)

print("\nb = Ax:")
print(b)

print("\nSolved x:")
print(solved_x)

print("\nSolutions match:")
print(np.allclose(original_x, solved_x))
```

3. Solve the following system.
8x+3y−2z=9
−4x+7y+5z==15
3x+4y−12z=35

```python

import numpy as np

A = np.array([
    [8, 3, -2],
    [-4, 7, 5],
    [3, 4, -12]
], dtype=float)

b = np.array([9, 15, 35], dtype=float)

solution = np.linalg.solve(A, b)

x, y, z = solution

print("x =", x)
print("y =", y)
print("z =", z)
```

np.dot() is used for dot product and matrix multiplication. It can also multiply a scalar with an array.

np.matmul() is mainly used for matrix multiplication. It does not allow scalar multiplication in the same way.

Task 5 - Plot f(x) = 4x^2 + 2


```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-5, 5, 200)
y = 4 * x**2 + 2
plt.plot(x, y, color='purple', label='f(x) = 4x^2 + 2')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Graph of f(x) = 4x^2 + 2')
plt.grid(True)
plt.legend()
plt.show()
```

Task 6 - Vector plotting and calculations
a. Create and plot three vectors in 3D

```python
U = np.array([1, 0.25, 0])
V = np.array([0, 1, 0.25])
W = np.array([0.25, 0, 1])
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
for vector, color, label in [(U, 'r', 'U'),
 (V, 'b', 'V'),
 (W, 'g', 'W')]:
 ax.quiver(0, 0, 0, vector[0], vector[1], vector[2],
 color=color, label=label)
ax.set_xlim(0, 1.2)
ax.set_ylim(0, 1.2)
ax.set_zlim(0, 1.2)
ax.set_xlabel('X'); ax.set_ylabel('Y'); ax.set_zlabel('Z')
ax.legend()
plt.show()
```

c. Function to calculate the magnitude of a vector

```python
def magnitude(vector):
 vector = np.asarray(vector, dtype=float)
 return np.sqrt(np.sum(vector**2))
U = np.array([1.5, 1.5, 2.0])
print(magnitude(U)) # 2.9154759474226504
```

d. Function to calculate direction cosines

```python
def direction_cosines(vector):
 vector = np.asarray(vector, dtype=float)
 length = magnitude(vector)
 if np.isclose(length, 0):
 raise ValueError('The zero vector has no direction cosines.')
 return vector / length
O = np.array([1.5, 1.5, 2.0])
print(direction_cosines(O))
# [0.51449576 0.51449576 0.68599434]
```

e. Function to calculate the scalar (dot) product


```python
def scalar_product(a, b):
 return np.dot(a, b)
a = np.array([1.5, 1.5, 2.0])
b = np.array([-1.5, 0.5, 2.0])
print(scalar_product(a, b)) # 2.5
```

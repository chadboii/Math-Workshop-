Task 1: Scatter graph with 10 random values

```python
import matplotlib.pyplot as plt
import random

x = []
y = []

for i in range(10):
    x.append(random.randint(0, 50))
    y.append(random.randint(0, 50))

plt.scatter(x, y)

plt.xlabel("Time taken in seconds")
plt.ylabel("Distance reached")
plt.title("Buzzer Game Results")

plt.show()
```

Task 1 with different colors

```python
import matplotlib.pyplot as plt
import random

x = []
y = []

for i in range(10):
    x.append(random.randint(0, 50))
    y.append(random.randint(0, 50))

colors = ["red", "green", "blue", "yellow", "pink", "orange", "purple", "brown", "cyan", "magenta"]

plt.scatter(x, y, c=colors)

plt.xlabel("Time taken in seconds")
plt.ylabel("Distance reached")
plt.title("Buzzer Game Results")

plt.show()
```

Explanation:
When c=colors is added, each scatter point gets a different color from the list. Since there are 10 points and 10 colors, every point is displayed in its own color.

Task 2: Animal population pie charts from 2017 to 2022

```python
import matplotlib.pyplot as plt

animals = ["Bears", "Dolphins", "Whales"]

population = {
    "2017": [8, 150, 80],
    "2018": [54, 77, 54],
    "2019": [93, 32, 100],
    "2020": [116, 11, 76],
    "2021": [137, 6, 93],
    "2022": [184, 1, 72]
}

for year, values in population.items():
    plt.pie(values, labels=animals, autopct="%1.1f%%")
    plt.title("Animal Population in " + year)
    plt.show()
```

Task 2b: Show all pie charts on a single pane using subplot

```python
import matplotlib.pyplot as plt

animals = ["Bears", "Dolphins", "Whales"]

population = {
    "2017": [8, 150, 80],
    "2018": [54, 77, 54],
    "2019": [93, 32, 100],
    "2020": [116, 11, 76],
    "2021": [137, 6, 93],
    "2022": [184, 1, 72]
}

plt.figure(figsize=(10, 8))

index = 1

for year, values in population.items():
    plt.subplot(2, 3, index)
    plt.pie(values, labels=animals, autopct="%1.1f%%")
    plt.title(year)
    index = index + 1

plt.suptitle("Animal Population from 2017 to 2022")
plt.tight_layout()
plt.show()
```

Example 4: Bar chart with grid explanation

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
data = [23, 45, 56, 78, 213]

plt.bar(x, data, width=0.5, color="maroon")

plt.title("A bar plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.grid(color="#95a5a6", linestyle="--", linewidth=2, axis="y", alpha=0.7)

plt.show()
```

Explanation:
The plt.grid() line adds horizontal grid lines only on the y-axis. The grid lines are grey, dashed, slightly thick, and partly transparent because of alpha=0.7.

Task 4f-A: Pie chart with explode

The workshop gives myexplode = [0.3, 0, 0, 0], but there are 5 car colours, so we should use 5 values:
